```python
import pandas as pd
from sklearn.model_selection import train_test_split
from category_encoders import TargetEncoder
```

```python
d={
    'col1':['A','B','B','D','B','A','C','C','A','A'],
    'col2':['S1','S2','S3','S3','S1','S3','S2','S2','S1','S3'],
    'col3':['M1','M2','M3','M2','M3','M2','M1','M3','M1','M2'],
    'col4':[10,30,20,40,10,30,50,20,30,60]
}
```

```python
df=pd.DataFrame(d)
```

```python
x=df.drop(columns='col4')
y=df['col4']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encode=TargetEncoder()
encode.fit_transform(xtrain,ytrain)
```

```python
encode.transform(xtest)
```

```python
encoder=TargetEncoder(cols=['col1'])#cols used to mention in which col we need to perform the transform
encoder.fit_transform(xtrain,ytrain)
```

```python
encoder.transform(xtest)
```
