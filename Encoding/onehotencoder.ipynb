```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
d={
    'cat_col1':['A','B','B','D','B','A','C','C','A','A'],
    'cat_col2':['S1','S2','S3','S3','S1','S3','S2','S2','S1','S3'],
    'lable':['class0','class1','class2','class3','class4','class5','class6','class7','class8','class9',]
}
```

```python
df=pd.DataFrame(d)
```

```python
x=df.drop(columns='lable')
y=df['lable']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encoder=OneHotEncoder()
encoder.fit_transform(xtrain).toarray() #toarray() is used for get the array
```

```python
encoder.get_feature_names_out() #to get created col name
```

```python
encoder=OneHotEncoder(sparse_output=False) #sparse_output=False used for get the array directely
encoder.fit_transform(xtrain)
```

```python
encoder.transform(xtest)
```
