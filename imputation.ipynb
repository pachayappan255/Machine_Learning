```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
d={
    'col1':[10,20,np.nan,20,10],
    'col2':[100,200,200,np.nan,300],
    'col3':['A',np.nan,'C','A','B']
}
```

```python
df=pd.DataFrame(d)
df
```

```python
imputer=SimpleImputer(strategy='constant',fill_value=0)
```

```python
imputer2=SimpleImputer(strategy='constant',fill_value='unknown values')
```

```python
imputer.fit_transform(df[['col1','col2']])
```

```python
imputer2.fit_transform(df[['col3']])
```

```python
imputer_mean=SimpleImputer(strategy='mean')
imputer_mean.fit_transform(df[['col1','col2']])
# imputer_mean.fit_transform(df[['col3']])  # ValueError: Cannot use mean strategy with non-numeric data:
```

```python
imputer_median=SimpleImputer(strategy='median')
imputer_median.fit_transform(df[['col1','col2']])
```

```python
imputer_mode=SimpleImputer(strategy='most_frequent')
imputer_mode.fit_transform(df[['col3']])
imputer_mode.fit_transform(df[['col1']]) # i used two fit trasform only for testing in real time don't use to fit transform for tow times
```
