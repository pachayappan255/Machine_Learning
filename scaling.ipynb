```python
x=df.drop(columns='Salary')
y=df['Salary']
```

```python
sns.boxplot(x)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
stand_scaling=StandardScaler()
minmax_scaling=MinMaxScaler()
```

```python
xtrain[['Experience_Years','Age','Hours_Per_Week']]=stand_scaling.fit_transform(xtrain[['Experience_Years','Age','Hours_Per_Week']])
```

```python
xtrain['Education_Level']=minmax_scaling.fit_transform(xtrain[['Education_Level']])
```

```python
xtest[['Experience_Years','Age','Hours_Per_Week']]=stand_scaling.transform(xtest[['Experience_Years','Age','Hours_Per_Week']])
```
