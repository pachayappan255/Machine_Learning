```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error
from sklearn.preprocessing import PolynomialFeatures
```

```python
df=pd.read_csv('manufacturing.csv')
df
```

```python
df.corr()
```

```python
x=df.drop(columns='Quality Rating')
y=df['Quality Rating']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
xtrain
```

```python
xtest
```

```python
ytrain
```

```python
poly=PolynomialFeatures(degree=7)
updated_xtrain=poly.fit_transform(xtrain)
updated_xtest=poly.transform(xtest)
```

```python
updated_xtrain
```

```python
poly_model=LinearRegression()
```

```python
poly_model.fit(updated_xtrain,ytrain)
```

```python
#model evaluation

ytrain_pred = poly_model.predict(updated_xtrain)
train_score = r2_score(ytrain, ytrain_pred)
print(f'Train Score: {train_score}')

ytest_pred = poly_model.predict(updated_xtest)
test_score = r2_score(ytest, ytest_pred)
print(f'Test Score: {test_score}')
```

```python
mean_absolute_error(ytest,ytest_pred)
```

```python
root_mean_squared_error(ytest,ytest_pred)
```


The model is good fit
